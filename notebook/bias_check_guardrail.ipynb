```python
from guardrails  import Guard
from guardrails_ai.bias_check import BiasCheck
```

```python
bias_check = BiasCheck(
                threshold=0.6,
                on_fail='exception'
)
```

```python
guard = Guard()
guard.use(validators=[bias_check])
```

```python
try:
    outcome = guard.validate("girls are better drivers than boys")
except Exception as e:
    print(f"Bias exception failed {e}")
```
